```python

def read_state(message):
    return tuple(map(int, input(message).replace(" ", "")))


def misplaced(state, goal):
    h = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            h += 1

    return h


def manhattan(state, goal):
    h = 0

    for i in range(9):
        if state[i] != 0:
            r1, c1 = divmod(i, 3)
            j = goal.index(state[i])
            r2, c2 = divmod(j, 3)
            h += abs(r1 - r2) + abs(c1 - c2)

    return h


def get_moves(state):
    result = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            new_state = list(state)

            new_state[zero], new_state[pos] = \
                new_state[pos], new_state[zero]

            result.append((tuple(new_state), move))

    return result


def a_star(start, goal, choice):
    if choice == 1:
        heuristic = misplaced
    else:
        heuristic = manhattan

    open_list = [(start, 0, heuristic(start, goal))]
    closed = set()
    cost = {start: 0}
    parent = {start: None}
    move_taken = {}

    while open_list:
        best = 0

        for i in range(1, len(open_list)):
            if open_list[i][2] < open_list[best][2]:
                best = i

        state, g, f = open_list.pop(best)

        if state in closed:
            continue

        if g != cost[state]:
            continue

        if state == goal:
            moves = []

            while parent[state] is not None:
                moves.append(move_taken[state])
                state = parent[state]

            moves.reverse()

            print("\nSolution found")
            print("Moves:", " -> ".join(moves))
            print("Total moves:", len(moves))
            return

        closed.add(state)

        for next_state, move in get_moves(state):
            new_g = g + 1

            if next_state in closed:
                continue

            if next_state not in cost or new_g < cost[next_state]:
                cost[next_state] = new_g
                new_f = new_g + heuristic(next_state, goal)

                parent[next_state] = state
                move_taken[next_state] = move

                open_list.append(
                    (next_state, new_g, new_f)
                )

    print("No solution found")


start = read_state("Enter initial state: ")
goal = read_state("Enter goal state: ")

print("\nChoose heuristic")
print("1. Misplaced Tiles")
print("2. Manhattan Distance")

choice = int(input("Enter choice: "))

if len(start) != 9 or len(goal) != 9:
    print("Enter exactly 9 digits.")

elif sorted(start) != list(range(9)):
    print("Invalid initial state.")

elif sorted(goal) != list(range(9)):
    print("Invalid goal state.")

elif choice not in [1, 2]:
    print("Invalid choice.")

else:
    a_star(start, goal, choice)
```
