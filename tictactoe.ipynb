```python
board = [[0 for i in range(3)] for i in range(3)]
print("Initial Board:-")

# Display the current Board
for i in range(3):
    for j in range(3):
        print(board[i][j], end=" ")
    print()

# If board is not filled
while not (all(x != 0 for i in board for x in i)):
    # take human input position
    print("\nEnter Board position to place X (Row 0-2 and Col 0-2):")
    r = int(input("Row: "))
    c = int(input("Col: "))
    
    # check input's validity
    if r < 0 or r > 2 or c < 0 or c > 2:
        print("Out of bounds! Choose rows and columns between 0 and 2.")
        continue
    if board[r][c] != 0:
        print("Invalid position. Try again")
        continue
    else:
        # place Human's X
        board[r][c] = 'X'
        
        # Display the Board after Human's turn
        print("\nBoard after your move:")
        for i in range(3):
            print(*(board[i]))

        # Check of 3X's in a row or column or diagonal, if yes, human wins
        human_wins = False
        # Check rows and columns
        for i in range(3):
            if board[i][0] == 'X' and board[i][1] == 'X' and board[i][2] == 'X':
                human_wins = True
            if board[0][i] == 'X' and board[1][i] == 'X' and board[2][i] == 'X':
                human_wins = True
        # Check diagonals
        if board[0][0] == 'X' and board[1][1] == 'X' and board[2][2] == 'X':
            human_wins = True
        if board[0][2] == 'X' and board[1][1] == 'X' and board[2][0] == 'X':
            human_wins = True

        if human_wins:
            print("\n Human wins! Game Over!")
            break

        # Check if board is filled (Draw check)
        if all(x != 0 for i in board for x in i):
            print("\nDRAW!\nGame Over!")
            break
        else:
            # Computer plays O
            # Find the first available empty spot         
            comp_played = False
            for i in range(3):
                for j in range(3):
                    if board[i][j] == 0:
                        board[i][j] = 'O'
                        comp_played = True
                        break
                if comp_played:
                    break

            # Display the Board after Computer's turn
            print("\nBoard after Computer's move:")
            for i in range(3):
                print(*(board[i]))

            # if 3 O's in a row or column or diagonal, Computer wins
            comp_wins = False
            # Check rows and columns
            for i in range(3):
                if board[i][0] == 'O' and board[i][1] == 'O' and board[i][2] == 'O':
                    comp_wins = True
                if board[0][i] == 'O' and board[1][i] == 'O' and board[2][i] == 'O':
                    comp_wins = True
            # Check diagonals
            if board[0][0] == 'O' and board[1][1] == 'O' and board[2][2] == 'O':
                comp_wins = True
            if board[0][2] == 'O' and board[1][1] == 'O' and board[2][0] == 'O':
                comp_wins = True

            if comp_wins:
                print("\nComputer wins! Game Over!")
                break
                
            # check if board filled
            if all(x != 0 for i in board for x in i):
                print("\nDRAW!\nGame Over!")
                break
            # repeat
```
