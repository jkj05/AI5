```python
import random

n = int(input("Enter the value of N: "))


board = [random.randint(0, n - 1) for _ in range(n)]


def cost(board):
    count = 0
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                count += 1
    return count


def display(board):
    print("Array:", board)
    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print("Cost:", cost(board))
    print()


print("\nINITIAL STATE")
display(board)

step = 1


while cost(board) != 0:
    neighbours = []

    print("STEP", step)


    for col in range(n):
        for row in range(n):
            if row != board[col]:
                new_board = board[:]
                new_board[col] = row
                neighbours.append(new_board)


    print("NEIGHBOURS AND THEIR COSTS:")
    for neighbour in neighbours:
        print(neighbour, " Cost =", cost(neighbour))


    best_board = min(neighbours, key=cost)
    print("\nSelected neighbour:", best_board)
    print("Selected cost:", cost(best_board))


    if cost(best_board) >= cost(board):
        print("\nLOCAL MINIMUM REACHED!")
        print("Restarting with a new random state...\n")
        board = [random.randint(0, n - 1) for _ in range(n)]
        print("NEW INITIAL STATE")
        display(board)
        continue

    board = best_board

    print("\nNEW STATE AFTER MOVING QUEEN:")
    display(board)

    step += 1


print("GOAL STATE REACHED!")
display(board)
```
