```python

import random
import math

N = 8

def conflicts(board):
    return sum(
        board[i] == board[j] or abs(board[i] - board[j]) == j - i
        for i in range(N) for j in range(i + 1, N)
    )

def queen_conflicts(board, col):
    return sum(
        board[col] == board[j] or abs(board[col] - board[j]) == abs(col - j)
        for j in range(N) if j != col
    )

def show(board):
    for r in range(N):
        print(" ".join("Q" if board[c] == r else "." for c in range(N)))

def solve():
    steps = 0
    restart = 0

    while True:
        restart += 1
        board = [random.randrange(N) for _ in range(N)]
        score = conflicts(board)
        temp = 2.5

        print(f"\nInitial State (Restart {restart}):")
        show(board)
        print("Initial Conflicts:", score)

        for _ in range(1000):
            if score == 0:
                break

            col = random.randrange(N)
            old = board[col]
            new = random.choice([r for r in range(N) if r != old])

            old_conf = queen_conflicts(board, col)
            board[col] = new
            new_conf = queen_conflicts(board, col)
            delta = new_conf - old_conf
            steps += 1

            accepted = (
                delta <= 0 or
                random.random() < math.exp(-delta / temp)
            )

            if accepted:
                score += delta
                result = "Accepted"
            else:
                board[col] = old
                result = "Rejected"

            print(
                f"Step {steps}: Column {col + 1}, "
                f"Row {old + 1} -> {new + 1}, "
                f"Conflicts: {score}, {result}"
            )

            temp = max(0.01, temp * 0.995)

        if score == 0:
            print("\nFinal State:")
            show(board)
            print("Total Steps:", steps)
            print("Total Restarts:", restart)
            print("Final Conflicts: 0")
            return

        print("Restarting search...")

solve()
```
